```python
import pandas as pd
```

## 1. Tokha last 365 days data cleaning

```python
df365 = pd.read_csv('Tokha_last365days.csv')
df365.info()
```

```python
df365.isnull().sum()
```

```python
df365.columns
```

```python
cleaned =df365.drop(columns=['country_iso', 'isMobile', 'isMonitor', 'datetimeUtc'], inplace=False)
cleaned
```

```python
cleaned = cleaned[cleaned['parameter']=='pm25']
cleaned
```

```python
# seperating date and time
cleaned['datetimeLocal']=pd.to_datetime(cleaned['datetimeLocal'])
cleaned['date']=cleaned['datetimeLocal'].dt.date
cleaned['time']=cleaned['datetimeLocal'].dt.strftime('%H:%M:%S')
cleaned
```

```python
cleaned.to_csv('cleaned_TokhaLast365days.csv', index=False)
```
