```python
import pandas as pd
```

## 1. Bagdol last 30 days data cleaning

```python
df30 = pd.read_csv('Bagdol_last30days.csv')
df30
```

```python
df30.info()
```

```python
df30.isnull().sum()
```

```python
df30.columns
```

```python
cleaned =df30.drop(columns=['country_iso', 'isMobile', 'isMonitor', 'datetimeUtc'], inplace=False)
cleaned
```

```python
df30['parameter'].unique()
```

```python
# cleaned = cleaned[cleaned['parameter']=='pm25']
# cleaned = cleaned[cleaned['parameter']=='temperature']
cleaned = cleaned[cleaned['parameter']=='relativehumidity']
cleaned
```

```python
cleaned['datetimeLocal'].unique()
```

```python
# seperating date and time
cleaned['datetimeLocal']=pd.to_datetime(cleaned['datetimeLocal'])
cleaned['date']=cleaned['datetimeLocal'].dt.date
cleaned['time']=cleaned['datetimeLocal'].dt.strftime('%H:%M:%S')
cleaned
```

```python
# cleaned['value'].unique
cleaned['unit'].unique
```

```python
cleaned.to_csv('cleaned_bagdolLast30days.csv', index=False)
```
